# Design device

Straight waveguide reference extending across the 26 mm width of the passive chip

```python
# Set up directories
device_name = 'A1-02_StraightReference'
device_dir = f"D:\\DATA\\ETL_PIC\\simulation\\NullingPIC_Gen2_sims\\{device_name}\\"
paths = device_paths(device_dir, device_name)

# Set up parameters
length = 26000 # length of passive chip

# Define circuit
c = rlib.Circuit(width=WIDTH)
guide = c.line((0.0, 0.0), (0.0, length), extended=True)
c.add_pathway([guide], launch=True)

# Export and visualize circuit
info = export_device(c, paths)
c.plot(labels=["straight reference"])                  # after the export, so the plot shows the domain as written
print(f"Wrote {paths['ind']} and {paths['gds']}; {length / SIM['Step_z']:.0f} propagation steps per wavelength")
```
