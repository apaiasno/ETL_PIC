# (d), (e) 2x2 MMI double input

- (b),(c) output injected in both inputs of (d), (e) respectively
- Phase offset applied in right input of both (d) and (e)
- Scan in wavelength and phase

```python
device_name = 'Device6a_I18_1x5Tri'
device_step_left, device_step_right = 'd', 'e'
device_dir = f"C:\\Users\\runner\\ActivePIC\\{device_name}\\"
save_dir_left = f"{device_dir}\\{device_name}_{device_step_left}_scan_Wav_Phase\\"
save_dir_right = f"{device_dir}\\{device_name}_{device_step_right}_scan_Wav_Phase\\"
fig_dir = f"C:\\Users\\runner\\ActivePIC\\device_figs\\{device_name}\\"
os.makedirs(fig_dir, exist_ok=True)
mon_dir = glob.glob(f"{save_dir_left}*.mon")
wavs = np.sort(np.array(list(set([float(path.split('Wav')[-1].split('_')[0].replace('p', '.')) for path in mon_dir]))))
phases = np.sort(np.array(list(set([float(path.split('Phase')[-1].split('.mon')[0].replace('p', '.')) for path in mon_dir]))))

output_A_left, output_B_left, output_A_right, output_B_right = np.nan*np.ones((len(wavs), len(phases))), np.nan*np.ones((len(wavs), len(phases))), np.nan*np.ones((len(wavs), len(phases))), np.nan*np.ones((len(wavs), len(phases)))
for i, wav in enumerate(wavs):
    wav_str = f"{wav:.3f}".replace('.', 'p')
    # norm_filename = norm_dir+f"StraightWaveguide_scan_Wav{wav_str}.mon"
    # norm_mon = np.genfromtxt(norm_filename, delimiter='', skip_header=5)
    # norm = norm_mon[-1, 1]
    for j, phase in enumerate(phases):
        phase_str = f"{phase:.3f}".replace('.', 'p')
        filename = save_dir_left+f"\\{device_name}_{device_step_left}_Wav{wav_str}_Phase{phase_str}.mon"
        sim = np.genfromtxt(filename, skip_header=5)       
        output_A_left[i,j] = sim[-1,1]
        output_B_left[i,j] = sim[-1,2]
    
        filename = save_dir_right+f"\\{device_name}_{device_step_right}_Wav{wav_str}_Phase{phase_str}.mon"
        sim = np.genfromtxt(filename, skip_header=5)       
        output_A_right[i,j] = sim[-1,1]
        output_B_right[i,j] = sim[-1,2]
output_A_left, output_B_left, output_A_right, output_B_right = np.array(output_A_left), np.array(output_B_left), np.array(output_A_right), np.array(output_B_right)
        
vmin, vmax = 0, 1

fig, ax = plt.subplots(2, 2, figsize=(12,10))
im = ax[0,0].pcolormesh(phases, wavs, output_A_left,  vmin=vmin, vmax=vmax)
ax[1,0].pcolormesh(phases, wavs, output_B_left,  vmin=vmin, vmax=vmax)
ax[0,1].pcolormesh(phases, wavs, output_A_right, vmin=vmin, vmax=vmax)
ax[1,1].pcolormesh(phases, wavs, output_B_right, vmin=vmin, vmax=vmax)
for i in range(2):
    ax[i,0].axvline(180, linestyle='dashed', color='white')
    ax[i,1].axvline(180, linestyle='dashed', color='white')
ax[0,0].set_title('(d)', fontsize=18)
ax[0,1].set_title('(e)', fontsize=18)
cbar = fig.colorbar(im, ax=ax.ravel().tolist(), orientation='vertical', fraction=0.03)
cbar.set_label('Normalized intensity', rotation=270, labelpad=20, fontsize=16)
fig.supxlabel("Relative Phase ($^\circ$)", fontsize=18)
fig.supylabel("Wavelength ($\\mathrm{\\mu}$m)", fontsize=18)

plt.savefig(f'{fig_dir}\\{device_name}_{device_step_left}_{device_step_right}.png',
            dpi=300, transparent=True, bbox_inches='tight')
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation, PillowWriter
import pandas as pd

# Read the dataset
df = pd.read_csv("Device6a_I18_1x5Tri_f_scan_Wav_I1_I2_phi/Device6a_I18_1x5Tri_f_scan_Wav_I1_I2_phi.dat")
wavs = np.sort(list(set(df["wav"])))

vmin, vmax = -40, 0

# ---- Use constrained layout ----
fig, ax = plt.subplots(1, 3, figsize=(18, 6), constrained_layout=True)
I1, I2 = np.sort(df["I1"].unique()), np.sort(df["I2"].unique())
dummy = np.zeros((len(np.sort(df["I2"].unique())),
                  len(np.sort(df["I1"].unique()))))

im0 = ax[0].pcolormesh(I1, I2, dummy, vmin=vmin, vmax=vmax, shading='auto')
im1 = ax[1].pcolormesh(I1, I2, dummy, vmin=vmin, vmax=vmax, shading='auto')
im2 = ax[2].pcolormesh(I1, I2, dummy, vmin=vmin, vmax=vmax, shading='auto')

ax[0].set_title('Output B', fontsize=18)
ax[1].set_title('Output C', fontsize=18)
ax[2].set_title('Output D', fontsize=18)

cbar = fig.colorbar(im0, ax=ax, orientation='vertical', fraction=0.035)
cbar.set_label('Attenuation (dB)', rotation=270, labelpad=20, fontsize=16)

fig.supxlabel("I1 ($\circ$)", fontsize=18)
fig.supylabel("I2 ($\circ$)", fontsize=18)

def update(frame):
    wav_fixed = wavs[frame]

    df_fixed = df[(df["wav"] == wav_fixed) & (df["phi"] == phi_fixed)]
    df_fixed = df_fixed.sort_values(["I2", "I1"])

    output_B = df_fixed.pivot(index="I2", columns="I1", values="output_B").values
    output_C = df_fixed.pivot(index="I2", columns="I1", values="output_C").values
    output_D = df_fixed.pivot(index="I2", columns="I1", values="output_D").values
    output_tot = output_B + output_C + output_D

    Z0 = 10*np.log10(output_B/output_tot)
    Z1 = 10*np.log10(output_C/output_tot)
    Z2 = 10*np.log10(output_D/output_tot)

    im0.set_array(Z0.ravel())
    im1.set_array(Z1.ravel())
    im2.set_array(Z2.ravel())

    fig.suptitle(f"Wavelength = {wav_fixed}", fontsize=20, color='r')

    return im0, im1, im2

anim = FuncAnimation(fig, update, frames=len(wavs), blit=False)
anim.save("wavelength_scan.gif", writer=PillowWriter(fps=2))

plt.close(fig)
```

```python
import matplotlib.cm as cm

# Read the dataset
df = pd.read_csv("Device6a_I18_1x5Tri_f_scan_Wav_I1_I2_phi/Device6a_I18_1x5Tri_f_scan_Wav_I1_I2_phi.dat")
wavs = np.sort(list(set(df["wav"])))

# ---- Use constrained layout ----
fig, ax = plt.subplots(1, 1, figsize=(10, 6), constrained_layout=True)
colormap_name = 'jet'
cmap = cm.get_cmap(colormap_name)
norm = plt.Normalize(vmin=np.min(wavs), vmax=np.max(wavs))

for wav_fixed in wavs:
    df_fixed = df[(df["wav"] == wav_fixed) & (df["phi"] == phi_fixed)]
    df_fixed = df_fixed.sort_values(["I2", "I1"])
    I1, I2 = np.sort(df_fixed["I1"].unique()), np.sort(df_fixed["I2"].unique())

    output_B = df_fixed.pivot(index="I2", columns="I1", values="output_B").values
    output_C = df_fixed.pivot(index="I2", columns="I1", values="output_C").values
    output_D = df_fixed.pivot(index="I2", columns="I1", values="output_D").values
    output_tot = output_B + output_C + output_D

    Z1 = 10*np.log10(np.min(output_C/output_tot, axis=1))
    
    im0 = ax.plot(I1, Z1, color=cmap(norm(wav_fixed)))  
    
    fig.supxlabel("I1 ($\circ$)", fontsize=18)
    fig.supylabel('Minimum attenuation (dB)', fontsize=18)

sm = cm.ScalarMappable(cmap=cmap, norm=norm)
sm.set_array([])  # required for matplotlib < 3.7
cbar = fig.colorbar(sm, ax=ax)
cbar.set_label('Wavelength ($\\mathrm{\\mu}$m)', rotation=270, labelpad=20, fontsize=16)
plt.show()
```
