```python
import io
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image

N_FRAMES = 20
FPS = 2

# XDAC channels loaded with PIC electrodes via the configuration pins
channels_load = np.array([26, 65, 24, 66, 71, 61, 62, 41, 63, 64, 42, 68, 13, 14, 44,
                          9, 10, 11, 12, 46, 45, 1, 2, 3, 4, 5, 6, 7, 8, 50, 49, 48, 47])

# Also connected to PIC electrodes, but driven by the Keysight rather than the XDAC
channels_highlight = np.array([43, 69])

fig, ax1 = plt.subplots(figsize=(10, 6))

ax1.axhline(0, color='C0', linestyle='dashed')
ax1.set_xlabel('Channel index')
ax1.set_ylabel('Voltage (V)', color='C0')
ax1.tick_params(axis='y', labelcolor='C0')

ax2 = ax1.twinx()
ax2.axhline(0, color='C1', linestyle='dashed')
ax2.set_ylabel('Current (mA)', color='C1', rotation=270, labelpad=25)
ax2.tick_params(axis='y', labelcolor='C1')

line_v, = ax1.plot([], [], '.', color='C0')
line_c, = ax2.plot([], [], '.', color='C1')

line_v_load, = ax1.plot([], [], 'o', markerfacecolor='C0', markeredgecolor='limegreen',
                        markersize=6, markeredgewidth=1.2, linestyle='none')
line_c_load, = ax2.plot([], [], 'o', markerfacecolor='C1', markeredgecolor='limegreen',
                        markersize=6, markeredgewidth=1.2, linestyle='none')

line_v_hl, = ax1.plot([], [], 'o', markerfacecolor='C0', markeredgecolor='purple',
                      markersize=7, markeredgewidth=1.6, linestyle='none', zorder=5)
line_c_hl, = ax2.plot([], [], 'o', markerfacecolor='C1', markeredgecolor='purple',
                      markersize=7, markeredgewidth=1.6, linestyle='none', zorder=5)

x = np.arange(1, XDACChannelMax + 1)
idx_load = channels_load - 1
idx_hl = channels_highlight - 1
ax1.set_xlim(0, XDACChannelMax + 1)

r0 = np.asarray(xdac.read_all())
ax1.set_ylim(min(r0[:, 0]) - 0.01, max(r0[:, 0]) + 0.01)
ax2.set_ylim(min(r0[:, 1]) - 0.01, max(r0[:, 1]) + 0.01)

def update(r):
    line_v.set_data(x, r[:, 0])
    line_c.set_data(x, r[:, 1])
    line_v_load.set_data(channels_load, r[idx_load, 0])
    line_c_load.set_data(channels_load, r[idx_load, 1])
    line_v_hl.set_data(channels_highlight, r[idx_hl, 0])
    line_c_hl.set_data(channels_highlight, r[idx_hl, 1])

# Freeze a padded tight bbox from one drawn frame, so every frame is cropped
# identically. bbox_inches='tight' recomputed per frame would vary the size.
update(r0)
fig.canvas.draw()
bbox = fig.get_tightbbox(fig.canvas.get_renderer()).padded(0.1)

frames = []
for _ in range(N_FRAMES):
    update(np.asarray(xdac.read_all()))

    buf = io.BytesIO()
    fig.savefig(buf, format='png', dpi=150, bbox_inches=bbox, facecolor='white')
    buf.seek(0)
    frames.append(Image.open(buf).convert('RGB'))

plt.close(fig)

frames[0].save('D:\\Active_PIC\\ControlCode\\CalibrationData\\CalibrationFigs\\XDAC_channels_off.gif',
               save_all=True, append_images=frames[1:], duration=int(1000/FPS), loop=0)
print('saved')
```

```python
import PowerMeterAcquire as pma

def connect_PM():
    pma.StartMeter()
    print('Connected to power meter.')

def disconnect_PM():
    pma.CloseMeter()
    print('Disconnected from power meter.')

def get_power(iter=10, std=False):
    powers=[]
    for it in range(iter):
        powers.append(pma.MeterReadPower()[0])
    if std == False:
        return np.average(powers)
    else:
        return [np.average(powers), np.std(powers)]
    
# tlx = TLX2.TLX() # Laser control object
# tlx.laser_on()

connect_PM() # Connect to power meter
```

```python
import numpy as np

def extinction_ratio(y, axis=None):
    """ER = max/min along the given axis (linear units)."""
    y_max = np.max(y, axis=axis)
    y_min = np.min(y, axis=axis)
    return y_max / y_min

def er_db(er_linear):
    return 10 * np.log10(er_linear)

print("=== 1D curves ===")
curves_1d = {
    'Monochromatic laser -> PM fiber -> power meter': pol_data[:, 1],
    'SLD -> PM fiber -> camera': outputs_SLD.flatten(),
    'Monochromatic laser -> PM fiber -> VOA -> power meter': pol_data_VOA[:, 1],
    'Monochromatic laser -> PM fiber -> polarization splitter -> Port 1 -> power meter': pol_data_split1[:, 1],
    'Monochromatic laser -> PM fiber -> polarization splitter -> Port 2 -> power meter': pol_data_split2[:, 1],
}
for label, y in curves_1d.items():
    er = extinction_ratio(y)
    print(f"{label}: ER = {er:.2f} ({er_db(er):.2f} dB)")

print()
print("=== 2D curves (one ER per column/channel) ===")
curves_2d = {
    'Monochromatic laser -> PM fiber -> VOA -> chip -> camera': outputs_mono_chip,
    'SLD -> PM fiber -> chip -> camera': outputs_SLD_chip,
}
for label, y2d in curves_2d.items():
    ers = extinction_ratio(y2d, axis=0)  # one ER per column/channel
    print(f"{label}:")
    for i, er in enumerate(ers):
        print(f"  Channel {i}: ER = {er:.2f} ({er_db(er):.2f} dB)")
```
